```python
# Trong Python, biến là các "nhãn" trỏ đến các đối tượng trong bộ nhớ, 
# thay vì là các "hộp" chứa dữ liệu có kích thước cố định như trong C.
# Điều này có nghĩa là khi bạn gán một giá trị mới cho một biến,
# bạn thực sự đang tạo ra một đối tượng mới và thay đổi nhãn để trỏ đến đối tượng đó, 
# thay vì thay đổi nội dung của đối tượng ban đầu.
a = 10
b = a  # b trỏ đến cùng một đối tượng với a
a = 20  # a bây giờ trỏ đến một đối tượng mới
print("a:", a)  # In ra 20
print("b:", b)  # In ra 10
```

```python
# Vì thế, một biến trong Python không có "kích thước" cố định;
# Điều này tạo ra sự linh hoạt lớn hơn trong việc quản lý bộ nhớ và xử lý dữ liệu.
# Một biến có thể trỏ đến các đối tượng có kích thước khác nhau mà không cần phải định nghĩa trước kích thước của nó.
# Một biến có thể trỏ đến một số nguyên, sau đó trỏ đến một chuỗi ký tự, 
# và sau đó trỏ đến một danh sách, tất cả trong cùng một chương trình.
# Ví dụ:

x = None  # x trỏ đến đối tượng None (không có giá trị)
print(x, type(x))

x = 42          # x trỏ đến một số nguyên (integer)
print(x, type(x))

x = 3.14        # bây giờ x trỏ đến một số thực (float)
print(x, type(x))

x = "Hello"     # bây giờ x trỏ đến một chuỗi ký tự (string)
print(x, type(x))

x = [1, 2, 3]   # bây giờ x trỏ đến một danh sách (có kích thước thay đổi)
print(x, type(x))

x= {1, 2, 3}    # bây giờ x trỏ đến một tập hợp (set)
print(x, type(x))

x = {'key': 'value'}  # bây giờ x trỏ đến một từ điển (dictionary)
print(x, type(x))

x = (4, 5)  # bây giờ x trỏ đến một tuple (bộ giá trị không thay đổi)
print(x, type(x))

# Tập hợp hỗn hợp (chỉ chứa các kiểu dữ liệu bất biến như số nguyên, số thực, chuỗi ký tự, tuple)
x = (1, 2.5, "text", (3, 4))
print(x, type(x))

# Điều này làm cho Python trở thành một ngôn ngữ rất linh hoạt và dễ sử dụng cho các lập trình viên.
```

```python
# Các phép toán trên biến trong Python cũng không bị giới hạn bởi kích thước cố định.
# Ví dụ, bạn có thể cộng hai chuỗi ký tự với nhau để tạo thành một chuỗi mới,
# hoặc bạn có thể nối hai danh sách để tạo thành một danh sách mới.
str1 = "Hello, "
str2 = "world!"
result_str = str1 + str2  # Nối hai chuỗi
print(result_str, type(result_str))
list1 = [1, 2, 3]
list2 = [4, 5, 6]
result_list = list1 + list2  # Nối hai danh sách
print(result_list, type(result_list))
```

```python

# Nối hai tuple
tuple1 = (1, 2)
tuple2 = (3, 4)
result_tuple = tuple1 + tuple2  # Nối hai tuple
print(result_tuple, type(result_tuple))
```

```python

# Nối hai từ điển
dict1 = {'a': 1, 'b': 2}
dict2 = {'c': 3, 'd': 4}
result_dict = {**dict1, **dict2}  # Nối hai từ điển
print(result_dict, type(result_dict))
```

```python

# Nối hai tập hợp
set1 = {1, 2, 3}
set2 = {4, 5, 6}
result_set = set1.union(set2)  # Nối hai tập hợp
print(result_set, type(result_set))
```

```python
# Cộng hai số nguyên
num1 = 10
num2 = 20
result_num = num1 + num2  # Cộng hai số nguyên
print(result_num, type(result_num))
```

```python
# Cộng hai số thực
float1 = 3.14
float2 = 2.86
result_float = float1 + float2  # Cộng hai số thực
print(result_float, type(result_float))
```

```python
# Cộng số nguyên và số thực
num1 = 10
float1 = 2.5
mixed_result = num1 + float1  # Cộng số nguyên và số thực
print(mixed_result, type(mixed_result))
```

```python
# Nối chuỗi và số nguyên
str1 = "Number: "
num1 = 42
str_num_result = str1 + str(num1)  # Nối chuỗi và số nguyên
print(str_num_result, type(str_num_result))
```

```python
# Nối chuỗi và số thực
str1 = "Value: "
float1 = 2.5
str_float_result = str1 + str(float1)  # Nối chuỗi và số thực
print(str_float_result, type(str_float_result))

```
