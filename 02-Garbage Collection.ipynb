```python
#Trong Python, trình thu gom rác sử dụng một cơ chế gọi là "đếm tham chiếu" để quản lý bộ nhớ.
#Mỗi đối tượng trong Python có một bộ đếm tham chiếu, theo dõi số lượng biến hoặc cấu trúc dữ liệu đang tham chiếu đến đối tượng đó.
a = [1, 2, 3]
b = a  # b tham chiếu đến cùng một đối tượng với a
a.append(4)  # Thay đổi đối tượng thông qua a
print("a:", a)  # In ra [1, 2, 3, 4]
print("b:", b)  # In ra [1, 2, 3, 4]
#Khi bộ đếm tham chiếu của một đối tượng giảm về 0 (tức là không còn biến nào tham chiếu đến nó),
#trình thu gom rác sẽ tự động giải phóng bộ nhớ mà đối tượng đó chiếm dụng.
#Điều này giúp ngăn ngừa rò rỉ bộ nhớ và đảm bảo rằng bộ nhớ được sử dụng hiệu quả.
```

```python
#Khi bạn tạo các biến và gán chúng cho các đối tượng, bộ đếm tham chiếu của các đối tượng đó sẽ tăng lên.
import sys #Module sys chứa các hàm và biến liên quan đến hệ thống và trình thông dịch Python
a = [1, 2, 3]  # Tạo một đối tượng danh sách và gán nhãn a cho nó
print("Bộ đếm tham chiếu của đối tượng danh sách:", sys.getrefcount(a))  # In ra số lượng tham chiếu: 2
#Tại sao lại là 2 mà không phải 1? Vì hàm getrefcount() cũng tạo một tham chiếu tạm thời đến đối tượng khi nó được gọi.
b = a          # b cũng trỏ đến cùng một đối tượng với a (nên bộ đếm tham chiếu tăng lên)
print("Bộ đếm tham chiếu của đối tượng danh sách khi b trỏ đến cùng một đối tượng với a:", sys.getrefcount(a))  # In ra số lượng tham chiếu
del a          # Xoá nhãn a (bộ đếm tham chiếu giảm xuống)
print("Bộ đếm tham chiếu của đối tượng danh sách sau khi xoá a:", sys.getrefcount(b))  # In ra số lượng tham chiếu
del b          # Xoá nhãn b (bộ đếm tham chiếu giảm xuống về 0, đối tượng được giải phóng)
#Khi cả hai biến a và b bị xoá, bộ đếm tham chiếu của đối tượng danh sách giảm về 0,
#và trình thu gom rác sẽ tự động giải phóng bộ nhớ mà đối tượng đó chiếm dụng.
```
