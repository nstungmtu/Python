```python
#Xử lý chuỗi trong Python
# Chuỗi trong Python là các đối tượng bất biến (immutable), có nghĩa là sau khi một chuỗi được tạo ra, bạn không thể thay đổi nội dung của nó. 
# Tuy nhiên, bạn có thể tạo ra các chuỗi mới dựa trên các chuỗi hiện có thông qua các phép toán và phương thức chuỗi.
s = "Hello"
s_upper = s.upper()  # Tạo chuỗi mới với tất cả chữ hoa
print(s)        # In ra "Hello" (chuỗi ban đầu không thay đổi)
print(s_upper)  # In ra "HELLO" (chuỗi mới)
```

```python
# Ví dụ khác về việc nối chuỗi
s1 = "Hello"
s2 = "World"
s3 = s1 + " " + s2  # Tạo chuỗi mới bằng cách nối s1 và s2
print(s3)  # In ra "Hello World"
```

```python
# Ví dụ về việc thay thế chuỗi
s = "Hello"
s4 = s.replace("H", "J")  # Tạo chuỗi mới với
# chữ 'H' được thay thế bằng 'J'
print(s4)  # In ra "Jello"
print(s)   # In ra "Hello" (chuỗi ban đầu không thay đổi)
# Như vậy, mặc dù bạn không thể thay đổi trực tiếp nội dung của một chuỗi trong Python,
# bạn có thể tạo ra các chuỗi mới dựa trên các chuỗi hiện có bằng cách sử dụng các phương thức và phép toán chuỗi.
```

```python
# Sử dung f-strings để định dạng chuỗi
name = "Alice"
age = 30
greeting = f"My name is {name} and I am {age} years old."
print(greeting)  # In ra: My name is Alice and I am 30 years old.
```

```python
# Xử lý chuỗi nâng cao với các phương thức chuỗi
text = "  Hello, World! Welcome to Python programming.  "
# Loại bỏ khoảng trắng ở đầu và cuối chuỗi
cleaned_text = text.strip()
print(cleaned_text)  # In ra: Hello, World! Welcome to Python programming.
```

```python

# Chuyển đổi chuỗi thành chữ thường
cleaned_text = "Hello, World! Welcome to Python programming."
lowercase_text = cleaned_text.lower()
print(lowercase_text)  # In ra: hello, world! welcome to python programming.
```

```python
# Tìm vị trí của một từ trong chuỗi
cleaned_text = "Hello, World! Welcome to Python programming."
position = cleaned_text.find("World")   
print(position)  # In ra: 7
```

```python

# Thay thế một từ trong chuỗi
cleaned_text = "Hello, World! Welcome to Python programming."
replaced_text = cleaned_text.replace("World", "Universe")
print(replaced_text)  # In ra: Hello, Universe! Welcome to Python programming.
```

```python

# Chia chuỗi thành danh sách các từ
cleaned_text = "Hello, World! Welcome to Python programming."
words = cleaned_text.split()
print(words)  # In ra: ['Hello,', 'World!', 'Welcome', 'to', 'Python', 'programming.']
```

```python

# Chia chuỗi thành danh sách từ với dấu phẩy làm dấu phân cách
cleaned_text = "Hello, World! Welcome to Python programming."
chars = cleaned_text.split(',')
print(chars)  # In ra: ['Hello', ' World! Welcome to Python programming.']
```

```python

# Chia chuỗi thành danh sách từ ký tự
cleaned_text = "Hello, World! Welcome to Python programming."
chars = list(cleaned_text)
print(chars)  # In ra: ['H', 'e', '... 'g', '.']
```

```python

# Tạo chuỗi từ danh sách các từ
words = ['Hello,', 'World!', 'Welcome', 'to', 'Python', 'programming.']
joined_text = ' '.join(words)
print(joined_text)  # In ra: Hello, World! Welcome to Python programming.
```
