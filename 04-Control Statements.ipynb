````python
# Lệnh rẽ nhánh mach-case (Python 3.10+)
day = 3
match day:
    case 1:
        print("Thứ Hai")
    case 2:
        print("Thứ Ba")
    case 3:
        print("Thứ Tư")
    case 4:
        print("Thứ Năm")
    case 5:
        print("Thứ Sáu")
    case 6:
        print("Thứ Bảy")
    case 7:
        print("Chủ Nhật")
    case _:
        print("Ngày không hợp lệ")
````

````python
# Lệnh rẽ nhánh match với nhiều giá trị
color = "xanh"
match color:
    case "đỏ" | "hồng":
        print("Màu đỏ hoặc hồng")
    case "xanh" | "lục":
        print("Màu xanh hoặc lục")
    case "vàng":
        print("Màu vàng")
    case _:
        print("Màu khác")
````

````python
# Lệnh rẽ nhánh match với nhiều giá trị
color = "đỏ"
hot_colors = ["đỏ","cam","vàng"]
match color:
    case color if color in hot_colors:
        print("Màu nóng")
    case _:
        print("Màu lạnh")
````

````python
# Lệnh rẽ nhánh match-case với nhiều mẫu
point = (0, 0) # Tọa độ điểm trong hệ trục tọa độ 2D
match point:
    case (0, 0):
        print("Điểm gốc")
    case (x, 0):
        print(f"Điểm trên trục x tại x={x}")
    case (0, y):
        print(f"Điểm trên trục y tại y={y}")
    case (x, y):
        print(f"Điểm tại tọa độ ({x}, {y})")
````

````python
# List comprehension
squares = [x**2 for x in range(10)]
print("Bình phương từ 0 đến 9:", squares)
````

````python
# List comprehension 2
even_numbers = [x for x in range(1,20) if x % 2 == 0]
print("Số chẵn từ 1 đến 20:", even_numbers)
````

````python
# List comprehension 3
grades = [85, 92, 78, 90, 88]
passed_grades = [grade for grade in grades if grade >= 80]
print("Điểm qua môn:", passed_grades)
````

Bài tập 1: Chuyển đổi vòng lặp cơ bản (Refactoring)  
Yêu cầu: Viết lại đoạn code sau sử dụng List Comprehension để code ngắn gọn hơn (chỉ còn 1 dòng).  
```
numbers = [1, 2, 3, 4, 5]
squares = []
for n in numbers:
    squares.append(n * n)
print(squares)
```

Bài tập 2: Làm phẳng danh sách (Flattening Matrix)  
Yêu cầu: Cho một ma trận 2 chiều (list chứa các list). Hãy sử dụng List Comprehension để "làm phẳng" (flatten) nó thành một list 1 chiều duy nhất.  
```
matrix = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]
# Viết code để output là: [1, 2, 3, 4, 5, 6, 7, 8, 9]
```

Bài tập 3: Trích xuất dữ liệu thực tế (Data Extraction)  
Yêu cầu: Bạn có một danh sách các dictionary chứa thông tin người dùng. Hãy tạo một danh sách chứa tên (name) của những người dùng đang hoạt động (active: True).  
```
users = [
    {"id": 1, "name": "Alice", "active": True},
    {"id": 2, "name": "Bob", "active": False},
    {"id": 3, "name": "Charlie", "active": True}
]
```


````python
#Bài tập 1: Chuyển đổi vòng lặp cơ bản (Refactoring)  
squares_dict = {x**2 for x in range(5)}
print("Từ điển bình phương:", squares_dict)
````

````python
#Bài tập 2: Làm phẳng danh sách (Flattening Matrix)  
matrix = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]
flattened = [num for row in matrix for num in row]
print("Danh sách đã làm phẳng:", flattened)
````

````python
#Bài tập 3: Trích xuất dữ liệu thực tế (Data Extraction)  
users = [
    {"id": 1, "name": "Alice", "active": True},
    {"id": 2, "name": "Bob", "active": False},
    {"id": 3, "name": "Charlie", "active": True}
]
active_users = [user["name"] for user in users if user["active"]]
print("Người dùng hoạt động:", active_users)
````
