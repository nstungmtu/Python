```python
#Truy xuất phần tử trong mảng
a=["Một","Hai",3,4,"Năm",6,7]
#In phần tử thứ nhất
print(f"Phần tử thứ nhất từ trái qua: {a[0]}")
#In phần tử cuối cùng
print(f"Phần tử cuối cùng: {a[-1]}")
#In 3 phần tử đầu tiên
print(f"3 phần tử đầu tiên: {a[:3]}")
#In 3 phần tử bắt đầu từ vị trí thứ 2
print(f"3 phần tử bắt đầu từ vị trí thứ 2: {a[1:4]}")
#In 3 phần tử cuối cùng
print(f"3 phần tử cuối cùng: {a[-3:]}")
```

```python
#List comprehension
a=["Một","Hai",3,4,"Năm",6,7]

#In các phần tử là số
print(f"Các phần tử là số {[i for i in a if type(i) == int]}")

#In các phần tử là số chẵn
print(f"Các phần tử là số chẵn: {[i for i in a if type(i) == int and i%2 == 0]}")
```

```python
#List comprehension
#Tạo mảng từ 0-9 với range
a=list(range(0,10,1))
print(a)

#Tạo mảng từ 9-0 với range
a=list(range(9,-1,-1))
print(a)
```

```python
#Thực thi hàm trong mảng
a=[]
[a.append(i) for i in range(10) if i%2 == 0]
print(a)
```
