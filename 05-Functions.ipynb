```python
# Khai báo hàm trong Python
def greet(name):
    return f"Xin chào, {name}!"

print(greet("Alice"))
```

```python
#Phạm vi sử dụng biến trong hàm
def outer_function():
    outer_var = "Biến bên ngoài"

    def inner_function():
        inner_var = "Biến bên trong"
        print(outer_var)  # Truy cập biến bên ngoài

    inner_function()
    # print(inner_var)  # Lỗi: inner_var không được định nghĩa trong phạm vi này
outer_function()
```

```python
#Hàm liên tục (Lambda Functions)
add = lambda x, y: x + y
print("Tổng:", add(3, 5))
```

```python
#Sử dụng lambda với hàm map và filter
numbers = [1, 2, 3, 4, 5]
squared = list(map(lambda x: x**2, numbers))
even_numbers = list(filter(lambda x: x % 2 == 0, numbers))
print("Bình phương:", squared)
print("Số chẵn:", even_numbers)
```

```python
#Hàm đệ quy
def factorial(n):
    if n == 0 or n == 1:
        return 1
    else:
        return n * factorial(n - 1)
print("Giai thừa của 5:", factorial(5))
```

```python
#Sử dụng hàm như đối số (Higher-Order Functions)
def apply_function(func, value):
    return func(value)
result = apply_function(lambda x: x**3, 4)
print("Kết quả áp dụng hàm:", result)
```

```python
#Hàm với tham số mặc định và tham số từ khóa
def greet_user(name, greeting="Xin chào"):
    return f"{greeting}, {name}!"
print(greet_user("Bob"))
print(greet_user("Bob", greeting="Chào mừng"))
```

```python
#Hàm với số lượng tham số không xác định
def sum_all(*args):
    return sum(args)
print("Tổng tất cả:", sum_all(1, 2, 3, 4, 5))
```

```python
#Hàm với số lượng tham số không xác định theo từ khóa
def print_info(**kwargs):
    for key, value in kwargs.items():
        print(f"{key}: {value}")
print_info(name="Alice", age=30, city="Hà Nội")
```

```python
#Sử dụng hàm lambda trong danh sách
functions = [lambda x: x + 1, lambda x: x * 2, lambda x: x ** 2]
for func in functions:
    print(func(5))
```
