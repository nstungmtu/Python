```python
#Cơ chế xử lý ngoai lệ trong Python
try:
    # Mã có thể gây ra ngoại lệ
    num = int(input("Nhập một số nguyên: "))
    result = 10 / num
    print(f"Kết quả: {result}")
except ValueError:
    # Xử lý ngoại lệ ValueError
    print("Lỗi: Vui lòng nhập một số nguyên hợp lệ.")
except ZeroDivisionError:
    # Xử lý ngoại lệ chia cho 0
    print("Lỗi: Không thể chia cho 0.")
except Exception as e:
    # Xử lý các ngoại lệ khác
    print(f"Lỗi không xác định: {e}")
```

```python
try:
    str = input("Nhập một chuỗi: ")
    index = int(input("Nhập một chỉ số để truy cập ký tự trong chuỗi: "))
    print(f"Ký tự tại vị trí {index} là: {str[index]}")
except ValueError:
    print("Lỗi: Vui lòng nhập một số nguyên hợp lệ cho chỉ số.")
except IndexError:
    print("Lỗi: Chỉ số nằm ngoài độ dài của chuỗi.")
except Exception as e:
    print(f"Lỗi không xác định: {e}")
    
```

```python
#Sử dụng try, except, else, finally trong Python
try:
    file = open("example.txt", "r")
    # Đọc nội dung tệp
    content = file.read()
except FileNotFoundError:
    # Xử lý ngoại lệ khi tệp không tồn tại
    print("Lỗi: Tệp không tồn tại.")
except Exception as e:
    # Xử lý các ngoại lệ khác
    print(f"Lỗi không xác định khi xử lý tệp: {e}")
else:
    # Mã chạy nếu không có ngoại lệ nào xảy ra
    print("Nội dung tệp:")
    print(content)
finally:
    # Mã luôn chạy, bất kể có ngoại lệ hay không
    try:
        # Đảm bảo tệp được đóng nếu nó đã được mở
        file.close()
    except NameError: # Nếu file chưa được mở
        # Không cần xử lý gì ở đây
        pass
    print("Kết thúc quá trình xử lý tệp.")

        
```

```python
#Phát một ngoại lệ tùy chỉnh trong Python
class NegativeNumberError(Exception):
    """Ngoại lệ tùy chỉnh cho số âm."""
    def __init__(self, number):
        self.number = number
        super().__init__(f"Lỗi: Số âm không được phép. Số đã nhập là {self.number}.")
    pass

def check_positive_number(num):
    """Kiểm tra nếu số là dương, nếu không thì phát ngoại lệ tùy chỉnh."""
    if num < 0:
        raise NegativeNumberError(num)
    else:
        print(f"Số {num} là số dương.")
        
try:
    number = int(input("Nhập một số: "))
    check_positive_number(number)
except NegativeNumberError as nne:
    print(nne)
except ValueError:
    print("Lỗi: Vui lòng nhập một số hợp lệ.")

```
