```python
# Sử dụng chuẩn PEP 8
# Lợi ích của việc sử dụng getter và setter để truy cập thuộc tính private:
# 1. Bảo vệ dữ liệu: Giúp kiểm soát cách thức truy cập và sửa đổi dữ liệu.
# 2. Tăng tính linh hoạt: Dễ dàng thay đổi cách thức lưu trữ dữ liệu mà không ảnh hưởng đến mã bên ngoài.
# 3. Kiểm tra dữ liệu: Cho phép thực hiện các kiểm tra khi thiết lập giá trị mới.
# 4. Dễ bảo trì: Mã dễ đọc và bảo trì hơn khi sử dụng getter và setter.

class Student:
    def __init__(self, name, age):
        self.__name = name
        self.__age = age
        self.__grades = []
        
    # Sử dụng getter và setter để truy cập thuộc tính private
    @property
    def name(self):
        return self.__name

    @name.setter
    def name(self, value):
        if not value or value.strip() == "":
            raise ValueError("Tên không được để trống.")
        self.__name = value
    
    @property
    def age(self):
        return self.__age
    
    @age.setter
    def age(self, value):
        if value < 0:
            raise ValueError("Tuổi không thể âm.")
        self.__age = value
    
    @property
    def grades(self):
        return self.__grades.copy()
    
    @grades.setter
    def grades(self, *values):
        for value in values:
            if not (0 <= value <= 10):
                raise ValueError("Điểm phải nằm trong khoảng từ 0 đến 10.")
        self.__grades = values.copy()

    def add_grade(self, grade):
        if 0 <= grade <= 10:
            self.__grades.append(grade)
        else:
            raise ValueError("Điểm phải nằm trong khoảng từ 0 đến 10.")

    def calculate_gpa(self):
        if not self.__grades:
            return 0.0
        return sum(self.__grades) / len(self.__grades)

    def __str__(self):
        return f"Student(Name: {self.__name}, Age: {self.__age}, GPA: {self.calculate_gpa():.2f})"
```
