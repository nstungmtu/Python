```python
#Đọc nội dung tập tin cấu hình config.ini vào một từ điển Python
import configparser
config = configparser.ConfigParser()
config.read('config.ini')
config_dict = {section: dict(config.items(section)) for section in config.sections()}
print(config_dict)
```

```python
#Ghi thông tin cáu hình từ một từ điển Python vào tập tin config.ini
#Đọc nội dung tập tin cấu hình config.ini vào một từ điển Python
import configparser
config = configparser.ConfigParser()
config.read('config.ini')
#Thay đổi thông tin trong từ điển
config['database']['host'] = '192.168.1.100'
#Ghi thông tin cấu hình từ một từ điển Python vào tập tin config.ini
with open('config.ini', 'w') as configfile:
    config.write(configfile)
print("Cấu hình đã được cập nhật.")
```

```python
#Đọc và ghi dữ liệu JSON (JavaScript Object Notation)
import json
#Ghi dữ liệu vào tập tin JSON
data = [{
    "name": "John Doe",
    "age": 30,
    "city": "New York"
}, {
    "name": "Jane Smith",
    "age": 25,
    "city": "Los Angeles"
}]
with open('data.json', 'w') as jsonfile:
    json.dump(data, jsonfile, indent=4) #Thêm thuộc tính indent để định dạng dễ đọc hơn
print("Dữ liệu đã được ghi vào data.json.")

#Đọc dữ liệu từ tập tin JSON
with open('data.json', 'r') as jsonfile:
    data_loaded = json.load(jsonfile)
print("Dữ liệu đã được đọc từ data.json:")
print(data_loaded)

#Thay đổi dữ liệu trong tập tin JSON
data_loaded.append({
    "name": "Alice Johnson",
    "age": 28,
    "city": "Chicago"
})
with open('data.json', 'w') as jsonfile:
    json.dump(data_loaded, jsonfile, indent=4)
print("Dữ liệu đã được cập nhật trong data.json.")

#Xem lại nội dung tập tin JSON sau khi cập nhật
with open('data.json', 'r') as jsonfile:
    updated_data = json.load(jsonfile)
print("Nội dung cập nhật của data.json:")
print(updated_data)
# --- IGNORE ---

```
