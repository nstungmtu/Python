```python
#Đọc và cập nhật dữ liệu từ cơ sở dữ liệu SQLite
import sqlite3
conn = sqlite3.connect('example.db')
c = conn.cursor()

#Đọc dữ liệu
print("Dữ liệu ban đầu:")
c.execute("SELECT * FROM students")
rows = c.fetchall()
for row in rows:
    print(row)
    
#Cập nhật dữ liệu
c.execute("UPDATE students SET age = ? WHERE id = ?", (22, 1))
conn.commit()

#Xem dữ liệu đã cập nhật
print("Dữ liệu sau khi cập nhật:")
c.execute("SELECT * FROM students")
rows = c.fetchall()
for row in rows:
    print(row)

#Đóng kết nối
conn.close()
```

```python
#Xóa dữ liệu
import sqlite3
conn = sqlite3.connect('example.db')
c = conn.cursor()

#Yêu cầu người dùng nhập ID của sinh viên cần xóa
row_to_delete = input("Nhập ID của sinh viên cần xóa: ")
try:
    row_to_delete = int(row_to_delete)
except ValueError:
    print("ID không hợp lệ. Vui lòng nhập một số nguyên.")
    conn.close()
    exit()
except sqlite3.Error as e:
    print(f"Lỗi cơ sở dữ liệu: {e}")
    conn.close()
    exit()
#Xác nhận sự tồn tại của ID trước khi xóa
c.execute("SELECT * FROM students WHERE id = ?", (row_to_delete,))
if c.fetchone() is None:
    print(f"Không tìm thấy sinh viên với ID {row_to_delete}.")
else:
    c.execute("DELETE FROM students WHERE id = ?", (row_to_delete,))
    print(f"Đã xóa sinh viên với ID {row_to_delete}.")
    conn.commit()

#Xem dữ liệu sau khi xóa
print("Dữ liệu sau khi xóa:")
c.execute("SELECT * FROM students")
rows = c.fetchall()
for row in rows:
    print(row)

#Đóng kết nối
conn.close()
```

```python
#Thay đổi thông tin bảng dữ liệu SQLite
import sqlite3
conn = sqlite3.connect('example.db')
c = conn.cursor()
#Thay đổi cột id cho phép tự động tăng
c.execute("ALTER TABLE students RENAME TO old_students")
c.execute('''CREATE TABLE students
                (id integer primary key autoincrement, name text, age integer)''')
c.execute('''INSERT INTO students (name, age)
                SELECT name, age FROM old_students''')
c.execute("DROP TABLE old_students")
conn.commit()
#Đóng kết nối
conn.close()
```

```python
#Chèn dữ liệu vào bảng SQLite từ input của người dùng
import sqlite3
conn = sqlite3.connect('example.db')
c = conn.cursor()
#Yêu cầu người dùng nhập thông tin sinh viên
name = input("Nhập thông tin sinh viên theo định dạng [Họ Tên][Khoảng trắng][Tuổi]: ")

#Tách tên và tuổi từ chuỗi đầu vào
#Tuổi là một số nguyên ở vị trí cuối cùng, phần còn lại là tên
parts = name.split()
age_str = parts[-1] #Lấy phần tử cuối cùng làm tuổi
name_str = " ".join(parts[:-1]) #Phần còn lại là tên
try:
    age = int(age_str)
except ValueError:
    print("Tuổi không hợp lệ. Vui lòng nhập một số nguyên.")
    conn.close()
    exit()

#Chèn dữ liệu vào bảng
c.execute("INSERT INTO students (name, age) VALUES (?, ?)", (name_str, age))
conn.commit()

#Kiểm tra dữ liệu đã chèn thành công chưa
c.execute("SELECT * FROM students WHERE name = ? AND age = ?", (name_str, age))
result = c.fetchone()
if result:
    print(f"Đã thêm sinh viên: {name_str}, Tuổi: {age}")
else:
    print("Không thể thêm sinh viên.")

#Liệt kê tất cả sinh viên trong bảng
print("Danh sách sinh viên hiện tại:")
c.execute("SELECT * FROM students")
rows = c.fetchall()
for row in rows:
    print(row)
    
#Đóng kết nối
conn.close()
```
