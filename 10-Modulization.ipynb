```python
#Modulization pattern
def main():
    print("Hello, World!") 

print(__name__)
if __name__ == "__main__": #Thuộc tính đặc biệt __name__ biểu thị tên của mô-đun hiện tại.
    main()
    
# Việc này chắcn chắn sẽ in ra "Hello, World!" khi bạn chạy tập lệnh này trực tiếp.
# Tuy nhiên, nếu bạn nhập tập lệnh này như một mô-đun trong một tập lệnh khác,
# nó sẽ không in ra gì cả vì khối mã bên trong if __name__ == "__main__": sẽ không được thực thi.
# Điều này rất hữu ích để tách biệt mã có thể tái sử dụng (khi được nhập như một mô-đun)
# và mã chỉ nên chạy khi tập lệnh được thực thi trực tiếp.
# Đây là một mẫu phổ biến trong Python để cho phép mã có thể tái sử dụng và đồng thời có thể chạy như một chương trình độc lập.

```
